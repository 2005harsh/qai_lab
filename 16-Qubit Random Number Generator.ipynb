```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(16, 16)
qc.h(range(16))                   # Hadamard on all qubits -> superposition
qc.measure(range(16), range(16))  # measure -> each qubit becomes 0 or 1

sim = AerSimulator()
counts = sim.run(qc, shots=1).result().get_counts()

bits = list(counts.keys())[0]
print("16-bit random number:", bits)
print("Decimal value:", int(bits, 2))
```
