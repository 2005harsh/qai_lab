```python
# Practical: Implementing 5-Qubit Quantum Fourier Transform (QFT)

# Step 1: Import required libraries
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
import numpy as np

# Step 2: Create a quantum circuit with 5 qubits
qc = QuantumCircuit(5)

# Step 3: Apply input state
# Initialize the first qubit to |1>
qc.x(0)

# Step 4: Apply Quantum Fourier Transform

# Hadamard gate on first qubit
qc.h(0)

# Controlled phase rotation gates
qc.cp(np.pi/2, 1, 0)
qc.cp(np.pi/4, 2, 0)
qc.cp(np.pi/8, 3, 0)
qc.cp(np.pi/16, 4, 0)

# Hadamard gate on second qubit
qc.h(1)

qc.cp(np.pi/2, 2, 1)
qc.cp(np.pi/4, 3, 1)
qc.cp(np.pi/8, 4, 1)

# Hadamard gate on third qubit
qc.h(2)

qc.cp(np.pi/2, 3, 2)
qc.cp(np.pi/4, 4, 2)

# Hadamard gate on fourth qubit
qc.h(3)

qc.cp(np.pi/2, 4, 3)

# Hadamard gate on fifth qubit
qc.h(4)

# Step 5: Swap qubits to reverse their order
qc.swap(0, 4)
qc.swap(1, 3)

# Step 6: Display the quantum circuit
print("5-Qubit Quantum Fourier Transform Circuit:")
print(qc.draw(output='text'))

# Step 7: Get the final quantum state
state = Statevector.from_instruction(qc)

print("\nFinal Statevector:")
print(state)

# Step 8: Display measurement probabilities
print("\nMeasurement Probabilities:")
print(state.probabilities_dict())
```
