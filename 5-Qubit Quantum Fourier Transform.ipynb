```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Step 1: Create quantum circuit with 3 qubits and 3 classical bits
qc = QuantumCircuit(3, 3)

# Step 2: Prepare the state to teleport |1>
qc.x(0)
qc.barrier()

# Step 3: Create an entangled Bell pair between q1 and q2
qc.h(1)
qc.cx(1, 2)
qc.barrier()

# Step 4: Alice performs her operations
qc.cx(0, 1)
qc.h(0)
qc.barrier()

# Step 5: Alice measures her two qubits
qc.measure(0, 0)
qc.measure(1, 1)

# Step 6: Bob applies correction operations
with qc.if_test((qc.clbits[1], 1)):
    qc.x(2)

with qc.if_test((qc.clbits[0], 1)):
    qc.z(2)

# Step 7: Bob measures the received qubit
qc.measure(2, 2)

# Step 8: Display the quantum circuit
print("Quantum Teleportation Circuit:")
print(qc.draw())

# Step 9: Initialize the simulator
sim = AerSimulator()

# Step 10: Execute the circuit
result = sim.run(qc, shots=1000).result()

# Step 11: Get measurement results
counts = result.get_counts()
print("Measurement Results:", counts)

# Step 12: Display histogram
plot_histogram(counts)
plt.show()
```
